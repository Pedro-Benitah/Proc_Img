```python
img_rgb = cv2.imread('facil.jpg',cv2.IMREAD_COLOR_RGB)
plt.imshow(img_rgb)
img_gray = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
```

```python
def calcHist(img):
  hist_arr = np.zeros(256, dtype=int)
  for px in img:
    hist_arr[px] += 1

  plt.bar(range(256), hist_arr)
```

```python
calcHist(img_gray)
```
