```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from urllib.request import urlretrieve
```

```python
# Meme público usado apenas durante a execução; a imagem não é versionada.
url_imagem = ('https://i.pinimg.com/736x/e4/e7/6e/'
              'e4e76e9d4467069b2367a12c41ef4175.jpg')
urlretrieve(url_imagem, 'imagem.png')
```

```python
img_bgr = cv2.imread('imagem.png')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(6, 3))
plt.imshow(img_gray, cmap='gray')
plt.title('Imagem em tons de cinza')
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.imshow(img_rgb)
plt.title('Imagem Original')
plt.axis('off')
plt.show()
```

```python
fig, ax = plt.subplots(1, 3, figsize=(6, 4))
ax[0].imshow(img_hsv[:, :, 0], cmap='gray')
ax[0].set_title('Canal H')
ax[1].imshow(img_hsv[:, :, 1], cmap='gray')
ax[1].set_title('Canal S')
ax[2].imshow(img_hsv[:, :, 2], cmap='gray')
ax[2].set_title('Canal V')
for eixo in ax:
    eixo.axis('off')
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(2, 2, figsize=(6, 6))
ax[0, 0].imshow(img_rgb)
ax[0, 0].set_title('RGB')
ax[0, 0].axis('off')
ax[0, 1].imshow(img_bgr)
ax[0, 1].set_title('BGR')
ax[0, 1].axis('off')
ax[1, 0].imshow(img_hsv)
ax[1, 0].set_title('HSV')
ax[1, 0].axis('off')
ax[1, 1].imshow(img_gray, cmap='gray')
ax[1, 1].set_title('GRAY')
ax[1, 1].axis('off')
plt.tight_layout()
plt.show()
```

```python
def quantizar(imagem, bits):
    niveis = 2 ** bits
    indices = np.floor(imagem.astype(np.float32) * niveis / 256.0)
    return np.round(indices * 255.0 / (niveis - 1)).astype(np.uint8)


bits_utilizados = [1, 2, 4, 8]
quantizadas = {bits: quantizar(img_gray, bits) for bits in bits_utilizados}

for bits, imagem_quantizada in quantizadas.items():
    assert np.unique(imagem_quantizada).size <= 2 ** bits
assert np.array_equal(quantizadas[8], img_gray)

fig, ax = plt.subplots(1, 5, figsize=(18, 4))
ax[0].imshow(img_gray, cmap='gray', vmin=0, vmax=255)
ax[0].set_title('Original em cinza')
ax[0].axis('off')

for eixo, bits in zip(ax[1:], bits_utilizados):
    eixo.imshow(quantizadas[bits], cmap='gray', vmin=0, vmax=255)
    eixo.set_title(f'{bits} bit(s)\n{2 ** bits} níveis')
    eixo.axis('off')

plt.tight_layout()
plt.savefig('quantizacao.png', dpi=300, bbox_inches='tight')
plt.show()
print('quantizacao.png salvo com sucesso.')
```

**Resposta:** À medida que reduzimos o número de bits, diminuímos a quantidade de níveis de intensidade disponíveis. Assim, detalhes sutis e transições suaves são perdidos, surgem faixas de tons (posterização) e regiões passam a ter intensidades mais uniformes. Com 8 bits, praticamente toda a variação original é preservada; com 1 bit, restam apenas dois níveis de intensidade.
